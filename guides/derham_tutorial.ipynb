# De Rham diagrams in five minutes

This executable guide compiles real `derham` commands with `tikzfigure`, then shows the result beside the LaTeX source. Download the notebook above to experiment. You need `pdflatex`, `derham.sty`, `tikzfigure`, `pymupdf`, and a Jupyter kernel.

```python
from pathlib import Path
import sys
sys.path.insert(0, str(Path.cwd().parent / "docs" / "tools"))
from derham_figure import show_derham
```

## Your first complex

A one-command three-dimensional de Rham complex uses generic spaces and differentials. Put the same command inside display math in a LaTeX article.

```python
show_derham(r"\derhamcomplex")
```

The vector-calculus preset labels the Sobolev spaces and arrows explicitly. The two-dimensional curl convention is scalar valued; use custom labels if your convention differs.

```python
show_derham(r"\derhamcomplex[dimension=3,notation=vector,spaces=sobolev]")
```

## Form notation and dimensions

For differential forms, `notation=forms` labels the exterior derivatives. Generic and form notation also support dimensions above three.

```python
show_derham(r"\derhamcomplex[dimension=4,notation=forms,spaces=sobolev]")
```

## A commuting FEEC diagram

The two rows share columns. Vertical labels default to degree-indexed projections. A drawn square communicates an intended relation; the package cannot prove that your chosen maps commute.

```python
show_derham(r"\derhamcommuting[dimension=3,notation=forms,spaces=sobolev]")
```

## Control every label

Space labels, horizontal operators on each row, and inter-row maps have separate keys. Lists cover all degrees, even if you later display only part of the complex.

```python
show_derham(r"\derhamcommuting[dimension=2,custom-spaces={X^0,X^1,X^2},custom-discrete-spaces={X_h^0,X_h^1,X_h^2},operators={D_0,D_1},discrete-operators={D_{h,0},D_{h,1}},maps={P_0,P_1,P_2}]")
```

## Physical and reference domains

With $F:\widehat\Omega\to\Omega$, the pullback $F^*$ goes from physical forms to reference forms. The preset follows that contravariant direction.

```python
show_derham(r"\derhammapped[dimension=2,notation=forms,spaces=sobolev]")
```

## Coefficients and degree-of-freedom maps

The coefficient preset relates discrete function spaces to coordinate spaces. It does not assume an orthonormal basis or identity mass matrix.

```python
show_derham(r"\derhamcoefficients[dimension=2]")
```

## Compose your own rows

`derhamdiagram` uses named rows. Connections can appear before the row declarations; names still resolve when the diagram renders.

```python
show_derham(r"\begin{derhamdiagram}[dimension=2]\derhamconnect[from=physical,to=reference,operator=F^*]\derhamconnect[from=reference,to=discrete,operator=\pi_h]\derhamrow[name=physical,kind=continuous]\derhamrow[name=reference,kind=reference]\derhamrow[name=discrete,kind=discrete]\end{derhamdiagram}")
```
